```python
import matplotlib.pyplot as plt
import numpy

# https://www.3blue1brown.com/lessons/prime-spirals/

def is_prime(x: int) -> bool:
    # print(f"Check {x} is a  prime?")
    if x == 1 or x == 0:
        return False
    
    for i in range(2, x//2+1):
        # print(f"testing {i}")
        if x % i == 0:
            return False

    return True


def get_prime_list(max: int) -> list[int]:

    prime_number_list = []

    for i in range(max):
        if is_prime(i) is True:
            prime_number_list.append(i)

    return prime_number_list
```

```python
# plot 10^4 points
prime_number_list = get_prime_list(10000)

# plot on polar plot
fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
fig.set_size_inches(10, 10)
ax.plot(prime_number_list, prime_number_list, '.b')
ax.set_title("Prime Spiral")
ax.set_axis_off()
plt.show()
```

```python
# plot 10^6 points
prime_number_list = get_prime_list(10**6)

# plot on polar plot
fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
fig.set_size_inches(10, 10)
ax.plot(prime_number_list, prime_number_list, ',b')
ax.set_title("Prime Spiral")
ax.set_axis_off()
plt.show()
```
