```python
import matplotlib.pyplot as plt
import numpy as np

# https://en.wikipedia.org/wiki/Logistic_map

# Equation: x[n+1] = mu * x[n] * (1 - x[n])
# x[n] is the population at time step n
# mu is the growth rate
#

def iterate(x, mu):
    return mu * x * (1 - x)
```

```python
# For a small growth rate, the population will stabilize at a fixed point.
mu = 2
x0 = 0.2
iterations = 10

x = []
xn = x0
for i in range(iterations):
    xn = iterate(xn, mu)
    x.append(xn)

plt.plot(range(iterations), x)
plt.title(f"mu = {mu}, x0 = {x0}")
plt.xlabel('Iteration')
plt.ylabel('Population')
```

```python
# For a medium growth rate, the population will take more time to stabilize.
mu = 2.9
x0 = 0.2
iterations = 50

x = []
xn = x0
for i in range(iterations):
    xn = iterate(xn, mu)
    x.append(xn)

plt.plot(range(iterations), x)
plt.title(f"mu = {mu}, x0 = {x0}")
plt.xlabel("Iteration")
plt.ylabel("Population")
```

```python
# For a bigger growth rate, the population will be periodic.
mu = 3.5
x0 = 0.2
iterations = 50

x = []
xn = x0
for i in range(iterations):
    xn = iterate(xn, mu)
    x.append(xn)

plt.plot(range(iterations), x)
plt.title(f"mu = {mu}, x0 = {x0}")
plt.xlabel("Iteration")
plt.ylabel("Population")
```

```python
# For a huge growth rate, the population will be chaos!
mu = 3.9
x0 = 0.2
iterations = 100

x = []
xn = x0
for i in range(iterations):
    xn = iterate(xn, mu)
    x.append(xn)

plt.plot(range(iterations), x)
plt.title(f"mu = {mu}, x0 = {x0}")
plt.xlabel("Iteration")
plt.ylabel("Population")
```

```python
# Let's see the result of different growth rate (mu) values
# The dots in the diagram means the population at a certain time step.


def bifurcation_diagram(
    x0: float,
    resolution: float,
    iteration: int,
    number_show: int,
    growth_rate_start: float = 0,
    growth_rate_end: float = 4.0,
    color: str = "k",
    alpha: float = 0.05
) -> None:
    """
    Args:
        x0 (float): initial population
        resolution (float): the resolution of growth rate (mu)
        iteration (int): the number of iterations to run for each growth rate
        number_show (int): the number of population values to show for each growth rate
        growth_rate_start (float): the starting value of growth rate (mu)
        growth_rate_end (float): the ending value of growth rate (mu)
        color (str): the color of the dots in the diagram, e.g. "b" for blue, "r" for red, "g" for green, etc.
        alpha (float): the transparency of the dots in the diagram, 0 is fully transparent, 1 is fully opaque
    """

    # growth rate
    growth_rate_slice = int((growth_rate_end - growth_rate_start) / resolution)
    mu = np.linspace(growth_rate_start, growth_rate_end, growth_rate_slice)


    # iterate the logistic map for each growth rate and store the results
    x = []
    for i in range(growth_rate_slice):
        x_iter = []

        xn = x0
        for _ in range(iteration-1):
            x_iter.append(xn)
            xn = iterate(xn, mu[i])
                        
        x.append(x_iter[-number_show:])

    # plot the bifurcation diagram
    plt.figure(figsize=(20, 12))
    plt.plot(mu, x, ","+color, alpha=0.05)
    plt.title('Bifurcation Diagram of the Logistic Map')
    plt.xlabel('Growth Rate (mu)')
    plt.ylabel('Population (x)')
```

```python
bifurcation_diagram(
    x0=0.7,
    resolution=0.0001,
    iteration=1000,
    number_show=200,
    growth_rate_start=2.0,
    growth_rate_end=4.0,
    alpha=0.8,
    color="b",
)
```
