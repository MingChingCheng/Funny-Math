```python
import matplotlib.pyplot as plt
import numpy as np

# https://en.wikipedia.org/wiki/Logistic_map


# starting point
x0 = 0.7

# growth rate
resolution = 0.0001
growth_rate_slice = int(1 / resolution)
mu = np.linspace(2.8, 4.0, growth_rate_slice)

# number of iteration
iteration = 1000

# number to plot (show after a lot of iterations)
number_show = 200

def iterate(x, mu):
    return mu * x * (1 - x)

# iterate the logistic map for each growth rate and store the results
x = []
for i in range(growth_rate_slice):
    x_iter = []

    xn = x0
    for j in range(iteration-1):
        x_iter.append(xn)
        xn = iterate(xn, mu[i])
        
        
    x.append(x_iter[-number_show:])


plt.figure(figsize=(20, 12))
plt.plot(mu, x, ',k', alpha=0.01)
plt.title('Bifurcation Diagram of the Logistic Map')
```
