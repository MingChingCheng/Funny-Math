```python
import matplotlib.pyplot as plt
import numpy as np

def int_to_strs(n: int) -> list[str]:
    """
    example:
      input: 7854
      output: ["7", "8", "5", "4"]
    """
    return [s for s in str(n)]


def strs_to_int(strs: list[str]) -> int:
    """
    example:
      input: ["7", "6", "9", "4"]
      output: 7694
    """
    n = 0
    for i in range(len(strs)):
        n += int(strs[i]) * (10**i)
    return n


def descending_order(n: int) -> int:
    """
    example:
      input: 5682
      output: 8652
    """
    strs = int_to_strs(n)
    strs.sort(reverse=False)
    return strs_to_int(strs)


def ascending_order(n: int) -> int:
    """
    example:
      input: 9832
      output: 2389
    """
    strs = int_to_strs(n)
    strs.sort(reverse=True)
    return strs_to_int(strs)

```

```python
def Kaprekar_routine(n: int) -> tuple[int, int]:
    """
    Args:
      n (int): input number

    Returns:
      step: how many iterations to reach stable
      constant: the stable number (Kaprekar constant)
    """
    
    n_old = n
    n_new = -1
    step = 0
    
    while True:
        step += 1
        a = descending_order(n_old)
        b = ascending_order(n_old)
        n_new = a - b

        # print(f"Iteration {step}: {a} - {b} = {n_new}")

        if n_old == n_new:
            return step - 1, n_new
        else:
            n_old, n_new = n_new, n_old

        if step > 10000:
            return -1, -1
```

```python
# Compute Kaprekar routine from 1000 to 9999
Kaprekar_routine_list = []

for n in range(1000, 10000):
    Kaprekar_routine_list.append(Kaprekar_routine(n)[0])

# Reshape 1D array into 2D array
Kaprekar_map = np.array(Kaprekar_routine_list).reshape((90, 100))

# plot image
plt.imshow(Kaprekar_map, cmap="YlGnBu")
plt.axis("off")
```
